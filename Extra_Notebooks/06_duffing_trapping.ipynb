# Duffing Oscillator: Trapping Region Flow
This notebook illustrates how a trapping region evolves under the flow of the Duffing oscillator.
We consider the system:
\begin{align}
\dot x &= y \\
\dot y &= x - x^3 - \gamma y
\end{align}
We visualize how a circular trapping region of radius 2 evolves over time.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from ipywidgets import interact, FloatSlider

gamma = 0.2

def duffing(t, z):
    x, y = z
    return [y, x - x**3 - gamma*y]

# initial boundary of trapping region: circle radius 2
theta = np.linspace(0, 2*np.pi, 300)
x0 = 2*np.cos(theta)
y0 = 2*np.sin(theta)
init_points = np.vstack([x0, y0]).T

def evolve_region(t):
    xs, ys = [], []
    for pt in init_points:
        sol = solve_ivp(duffing, [0, t], pt, rtol=1e-6, atol=1e-9)
        xs.append(sol.y[0, -1])
        ys.append(sol.y[1, -1])
    return np.array(xs), np.array(ys)

def plot_region(t=0.0):
    xs, ys = evolve_region(t)
    plt.figure(figsize=(6,6))
    plt.plot(xs, ys)
    circle = plt.Circle((0,0), 2, color='gray', fill=False, linestyle='--')
    plt.gca().add_patch(circle)
    plt.xlim(-3,3)
    plt.ylim(-3,3)
    plt.gca().set_aspect('equal', 'box')
    plt.title(f'Evolved region at t={t:.2f}')
    plt.show()

interact(plot_region, t=FloatSlider(min=0,max=50,step=0.1,value=0));
```

- As the flow evolves, the image of the trapping region is stretched and folded, and asymptotically it collapses onto the two stable equilibria. The limiting attracting set is therefore a set of measure zero.

- For any finite time, the evolved region still encloses both stable equilibria and the unstable equilibrum at (0.0), and becomes elongated along the stable manifold of the saddle at the origin. The geometry can become extremely thin and convoluted.

- Numerical integration with a finite number of boundary points does not resolve these thin structures, especially near the separatrix.

- The long-term limiting set is attracting, but it is not a _minimal_ attractor. Each of the two stable equilibria is a true (minimal) attractor with its own basin of attraction; their union is an attracting set but not an attractor in the strict sense.
