# Excitability in the Shifted Van der Pol Oscillator

```python
# ==============================================================
# Imports and setup
# ==============================================================
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from ipywidgets import interact, FloatSlider

plt.rcParams['figure.figsize'] = (6, 4)
plt.rcParams['axes.grid'] = True

# ==============================================================
# Define the shifted Van der Pol system with stochastic forcing
# ==============================================================
def vdp_excitable(t, state, alpha, beta):
    x, y = state
    dxdt = y + beta
    dydt = alpha * (y - y**3 / 3 - x)
    return [dxdt, dydt]

def simulate_excitable(beta=1.01, alpha=30, sigma=0.5, t_max=10):
    dt = 0.001
    t = np.arange(0, t_max, dt)
    x = np.zeros_like(t)
    y = np.zeros_like(t)
    x[0], y[0] = 0.0, 0.0

    # Euler-Maruyama integration
    for i in range(1, len(t)):
        dx, dy = vdp_excitable(t[i-1], [x[i-1], y[i-1]], alpha, beta)
        x[i] = x[i-1] + dx * dt + sigma * np.sqrt(dt) * np.random.randn()
        y[i] = y[i-1] + dy * dt

    # Plot time series
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    axes[0].plot(t, x, lw=0.8)
    axes[0].set_xlabel("Time")
    axes[0].set_ylabel("x(t)")
    axes[0].set_title("Time series (excitable response)")

    # Phase portrait
    axes[1].plot(x, y, lw=0.8)
    axes[1].set_xlabel("x")
    axes[1].set_ylabel("y")
    axes[1].set_title("Phase portrait")
    plt.tight_layout()
    plt.show()
    
# ==============================================================

# Interactive exploration

# ==============================================================

interact(simulate_excitable,
beta=FloatSlider(value=1.01, min=1.0, max=1.1, step=0.002, description="β"),
sigma=FloatSlider(value=0.5, min=0.0, max=0.5, step=0.005, description="σ"),
t_max=FloatSlider(value=10, min=2, max=50, step=1, description="t_max"),
alpha=FloatSlider(value=30, min=10, max=50, step=1, description="α"));
```
