We use the powerful package [DynamicalSystems.jl](https://juliadynamics.github.io/DynamicalSystems.jl) package for genereating 1 million points with the Hénon map and show the graphical result.  

```julia
using DynamicalSystems
using PredefinedDynamicalSystems
using StaticArrays # only necessary when defining a system

he = PredefinedDynamicalSystems.henon([0.,0.0]; a = 1.4, b = 0.3)
```

```julia
traj = trajectory(he, 100000);
```

```julia
using GR
shade(traj[1][:,1],traj[1][:,2], colormap=-2, xlabel="x", ylabel="y")
```
